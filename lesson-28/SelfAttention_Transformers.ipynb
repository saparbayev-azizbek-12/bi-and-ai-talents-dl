```python
import torch.nn as nn
class SelfAttention_v1(nn.Module):
    def __init__(self, d_in, d_out):
        super().__init__()
        self.W_query = nn.Parameter(torch.rand(d_in, d_out))
        self.W_key = nn.Parameter(torch.rand(d_in, d_out))
        self.W_value = nn.Parameter(torch.rand(d_in, d_out))

    def forward(self, x):
        keys = x @ self.W_key
        queries = x @ self.W_query
        values = x @ self.W_value
        attn_scores = queries @ keys.T
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )
        context_vec = attn_weights @ values
        return context_vec
```

```python
import torch
from torch import nn

class SelfAttention_v3(nn.Module):
    def __init__(self, d_in, d_out, bias=False):
        super().__init__()
        self.w_key = nn.Linear(d_in, d_out, bias=bias)
        self.w_query = nn.Linear(d_in, d_out, bias=bias)
        self.w_value = nn.Linear(d_in, d_out, bias=bias)

    def forward(self, x):
        key = self.w_key(x)     # shape [n, d_out]
        query = self.w_query(x) # shape [n, d_out]
        value = self.w_value(x) # shape [n, d_out]

        att_scores = query @ key.T
        att_weights = torch.softmax(att_scores / key.size(-1)**0.5, dim=1)
        context_vector = att_weights @ value
        return context_vector
```

Multi head attention

```python
class CausalAttention_v1(nn.Module):
    def __init__(self, d_in, d_out, context_length, dropout_p, bias=False):
        super().__init__()
        self.w_key = nn.Linear(d_in, d_out, bias=bias)
        self.w_query = nn.Linear(d_in, d_out, bias=bias)
        self.w_value = nn.Linear(d_in, d_out, bias=bias)
        self.dropout = nn.Dropout(dropout_p)
        self.register_buffer(
            "mask",
            torch.triu(context_length, context_length, diagonal=1)
        )


    def forward(self, x):
        T = x.size(-2)
        key = self.w_key(x)
        query = self.w_query(x)
        value = self.w_value(x)

        att_scores = query @ key.transpose(-1, -2)
        att_scores.masked_fill_(
            self.mask.bool()[:T, T:], -torch.inf
        )
        att_weights = torch.softmax(att_scores / key.size(-1)**0.5, dim=1)
        att_weights = self.dropout(att_weights)
        context_vector = att_weights @ value
        return context_vector
```

```python
class MultiHeadAttention_v1(nn.Module):
    def __init__(self, d_in, d_out, context_length, dropout_p, num_heads):
        self.heads = nn.ModuleList(
            [
                CausalAttention_v1(d_in, d_out, context_length, dropout_p)
                for _ in range(num_heads)
            ]
        )

    def forward(self):
        torch.cat([head(x) for head in self.heads], dim=-1)
```

```python
class MultiHeadAttention_v2(nn.Module):
    def __init__(self, d_modul, num_heads, dropout, context_length, qkv_bias=False):
        super().__init__()
        if d_modul % num_heads != 0:
            raise "d_modul must be divisable num_heads"
        self.d_modul = d_modul
        self.num_heads = num_heads
        self.head_dim = d_modul // num_heads
        self.W_key = nn.Linear(d_modul, d_modul, bias=qkv_bias)
        self.W_value = nn.Linear(d_modul, d_modul, bias=qkv_bias)
        self.W_query = nn.Linear(d_modul, d_modul, bias=qkv_bias)
        self.register_buffer(
            "mask",
            torch.triu(torch.ones(context_length, context_length), diagonal=1)
        )
        self.output = nn.Linear(d_modul, d_modul)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [B, T, d_modul]
        B, T, d_modul = x.shape
        keys = self.W_key(x).view(B, T, self.num_heads, self.head_dim) # shape: [B, T, 4, 196]
        values = self.W_value(x).view(B, T, self.num_heads, self.head_dim)
        queries = self.W_query(x).view(B, T, self.num_heads, self.head_dim)

        keys = keys.transpose(1, 2) # shape: [B, 4, T, 196]
        values = values.transpose(1, 2)
        queries = queries.transpose(1, 2)

        att_scores = queries @ keys.transpose(-1, -2)
        mask_bool = self.mask.bool()[:T, :T]
        att_scores.masked_fill_(mask_bool, -torch.inf)
        att_weights = torch.softmax(
            att_scores / keys.size(-1)**0.5, dim=-1
        )
        att_weights = self.dropout(att_weights)
        context_vec = (att_weights @ values).transpose(1, 2)
        context_vec = context_vec.contiguous().view(B, T, self.d_modul)
        context_vec = self.output(context_vec)

        return context_vec
```

```python
x = torch.randn(4, 10, 784)
```

```python
d_modul = 784
num_heads = 4
dropout = 0.3
context_length = 10

model = MultiHeadAttention_v2(d_modul, num_heads, dropout, context_length)
out = model.forward(x)
```
